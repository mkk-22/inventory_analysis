## Import Libraries and Data

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/yallamotor_cars.csv")
df.head()
```

## Preprocessing

```python
print(df.shape) # 2061 rows, 35 columns
display(df.head()) 

# should remove price_text, price_valid_until, currency, availability, condition, source (they are always the same value)
df = df.drop(columns=["price_text", "price_valid_until", "currency", "availability", "condition", "source"])
df.head()
```

```python
# convert to correct datatypes
print(df.info()) # cols and their dtypes

# convert datetime col
df["scraped_at"] = pd.to_datetime(df["scraped_at"])

# numeric cols that get interpreted as "object" because of nulls
columns = ["number_of_doors", "seating_capacity", "engine_displacement_cc"]
for col in columns:
    df[col] = df[col].astype("Int64")

# boolean cols
df["certified"] = df["certified"].map({"Certified": True, pd.NA: pd.NA}).astype("boolean")
df["featured"] = df["featured"].replace({"True": True, "False": False}).astype("boolean")
```

```python
# handle duplicates (none found)
print("Duplicate listing IDs:", df["listing_id"].duplicated().sum())
print("Duplicate URLs:", df["url"].duplicated().sum())
print("Fully duplicated rows:", df.duplicated().sum())
```
